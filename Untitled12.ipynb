```python

# ==========================================
# TUGAS CNN PENGOLAHAN GAMBAR
# Dataset: CIFAR-10
# Library: TensorFlow / Keras
# ==========================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import cifar10

# 1. Memuat dataset CIFAR-10
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

print("Jumlah data training:", x_train.shape)
print("Jumlah data testing :", x_test.shape)

# 2. Normalisasi gambar
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# 3. Menentukan nama kelas gambar
class_names = [
    "Pesawat", "Mobil", "Burung", "Kucing", "Rusa",
    "Anjing", "Katak", "Kuda", "Kapal", "Truk"
]

# 4. Menampilkan contoh gambar
plt.figure(figsize=(8, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[int(y_train[i][0])])
    plt.axis("off")

plt.tight_layout()
plt.show()

# 5. Membuat model CNN
model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),

    layers.Conv2D(
        32, (3, 3), activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(
        64, (3, 3), activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

# 6. Menyiapkan model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 7. Menampilkan arsitektur model
model.summary()

# 8. Melatih model CNN
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2
)

# 9. Menguji model
test_loss, test_accuracy = model.evaluate(
    x_test, y_test, verbose=0
)

print("Test Loss:", test_loss)
print("Test Accuracy:", round(test_accuracy * 100, 2), "%")

# 10. Memprediksi gambar
predictions = model.predict(x_test[:10])
predicted_classes = np.argmax(predictions, axis=1)

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i])

    actual = class_names[int(y_test[i][0])]
    predicted = class_names[predicted_classes[i]]

    plt.title(
        f"Prediksi: {predicted}\nAsli: {actual}",
        fontsize=9
    )
    plt.axis("off")

plt.tight_layout()
plt.show()

# 11. Menyimpan model
model.save("model_cnn_cifar10.keras")

print("Model berhasil disimpan.")
```
